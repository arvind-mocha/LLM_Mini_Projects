```python
import sounddevice as sd
import soundfile as sf
import subprocess

duration = 10          # seconds to record
sample_rate = 44_100
channels = 1
print(duration * sample_rate)
print("Recording...")
audio = sd.rec(
    int(duration * sample_rate),
    samplerate=sample_rate,
    channels=channels,
    dtype="float32"
)
sd.wait()
print("Finished recording.")

# Save a temporary WAV file
sf.write("recording.wav", audio, sample_rate)

# Convert WAV to MP3 using FFmpeg
subprocess.run([
    "ffmpeg",
    "-y",
    "-i", "recording.wav",
    "recording.mp3"
], check=True)

print("Saved as recording.mp3")
```

```python
from transformers import pipeline
from langchain_core.runnables import RunnableLambda

# Step 1: Load Whisper for audio transcription
transcriber = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-small",
    device="mps"
)

# Step 2: Load a chat model for summarization.
# Transformers v5 removed the old `summarization` pipeline.
summarizer = pipeline(
    "text-generation",
    model="Qwen/Qwen3-1.7B",
    device="mps"
)

# Step 3: Define transcription function
def transcribe_audio(audio_path):
    result = transcriber(
        audio_path,
        generate_kwargs={"language": "english"}
    )
    return result["text"]

# Step 4: Ask the chat model to summarize the transcript
def summarize_text(text):
    messages = [
        {
            "role": "user",
            "content": f"Summarize this transcript in 3 short bullet points:\n\n{text}"
        }
    ]
    result = summarizer(messages, max_new_tokens=130, do_sample=False)
    return result[0]["generated_text"][-1]["content"]

# Step 5: Create LangChain model chain
chain = (
    RunnableLambda(transcribe_audio)
    | RunnableLambda(summarize_text)
)

# Step 6: Execute the chain
summary = chain.invoke("recording.mp3")

print("Summary:", summary)
```
